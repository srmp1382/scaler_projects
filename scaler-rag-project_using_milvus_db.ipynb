```python
def context_retriever(user_question:str):
    question_embedding=embed_model.embed_query(user_question)
    ans=client.search(collection_name='chunks_collection', data=[question_embedding], limit=1, output_fields=['id','text'], search_parameters={'metric_type':'COSINE'})
    ans_df=pd.DataFrame(ans).T
    ans_text=result2_text=[ans_df.iloc[i].values[0]['entity']['text'] for i in range(ans_df.shape[0])]
    #answer=" ".join(i for sub_list in ans_text for i in sub_list)
    return ans_text


def answer_generator(user_question:str,context:str):
    llm=ChatGroq(model='groq/compound-mini', temperature=0.2, api_key=groq_key)
    messages=[SystemMessage(content=f'''You are helpful assistant and are NOT to use your own knowledge or the internet.
                                        You will answer questions only from the context {context}.'''),
              HumanMessage(content=user_question)]
    prompt=ChatPromptTemplate.from_messages(messages)
    parser=StrOutputParser()
    chain=prompt|llm|parser
    return chain.invoke({'context':context,'user_question':user_question})


def run_rag(user_question: str):
    context=context_retriever(user_question)
    answers=answer_generator(user_question, context)
    return {'context':context,'answer':answers}
```
